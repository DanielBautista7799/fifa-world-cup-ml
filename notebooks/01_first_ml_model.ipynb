```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
rating_difference = np.array([
    -200,
    -150,
    -100,
    -50,
    0,
    50,
    100,
    150,
    200,
])

goal_difference = np.array([
    -3,
    -2,
    -1,
    -1,
    0,
    1,
    1,
    2,
    3,
])
```

```python
plt.scatter(rating_difference,goal_difference)
plt.xlabel("Rating Difference")
plt.ylabel("Goal Difference")
plt.title("Rating difference vs goal difference")
plt.show()
```

```python
weight = 0.0
bias = 0.0
prediction = weight * rating_difference + bias
print(prediction)

errors = goal_difference - prediction
squared_errors = (errors)**2
Mean_Squared_Error = np.mean(squared_errors)
print(errors)
print(squared_errors)
print(Mean_Squared_Error)
```

```python
n = len(rating_difference)

weight_gradient = (-2/n) * np.sum(rating_difference*(goal_difference - prediction))
print(weight_gradient)

learning_rate = 0.00001
weight = weight - (learning_rate*weight_gradient)
print(weight)

weight = 0.0
bias = 0.0
learning_rate = 0.00001

for epoch in range(1000):
    predictions = weight * rating_difference + bias
    errors = goal_difference - predictions 
    
    weight_gradient = (-2/n) * np.sum(rating_difference * errors)
    weight = weight - learning_rate * weight_gradient
    bias_gradient = (2/n) * np.sum(errors)
    bias = bias -learning_rate * bias_gradient
    
print(weight)
print(bias)

plt.scatter(rating_difference, goal_difference)

plt.plot(rating_difference, predictions)

plt.xlabel("Rating Difference")
plt.ylabel("Goal Difference")
plt.title("Trained Linear Regression Model")

plt.show()
```
